# Heterogeneous AI: Distributed CPU/GPU Execution

This notebook demonstrates a **Heterogeneous AI** pattern:
- **CPU (Intel Xeon on Colab)**: Handles lightweight, low-latency, or highly secure tasks like orchestrating tool calls, processing structured data, and running verification logic.
- **GPU (OpenRouter / Nemotron)**: Handles the heavy lifting of LLM prefill and token decoding (text generation).

### Step 2: CPU-Bound Tools & Data Processing
Here, we define local functions that run entirely on the **Intel Xeon CPU** in our Colab runtime. These simulate data processing and calculations that would be slow or impossible to run directly inside an LLM's context.

```python
import json
import pandas as pd
import platform

def cpu_print_specs():
    """Utility to display the local CPU specifications running this tool."""
    print(f"[CPU Exec] Hardware Processor: {platform.processor()}")
    print(f"[CPU Exec] System Platform: {platform.platform()}")

def cpu_calculate_financial_metrics(revenue_data_json: str) -> str:
    """
    Runs structured data processing and statistical analysis on the local Intel Xeon CPU.
    """
    cpu_print_specs()
    print("[CPU Exec] Processing and verifying financial payload...")

    try:
        data = json.loads(revenue_data_json)
        df = pd.DataFrame(data)

        # Local CPU calculations
        df['Profit_Margin'] = (df['Net_Profit'] / df['Revenue']) * 100
        avg_margin = df['Profit_Margin'].mean()
        total_revenue = df['Revenue'].sum()

        result = {
            "status": "success",
            "total_revenue": float(total_revenue),
            "average_profit_margin_percent": round(float(avg_margin), 2),
            "processed_rows": len(df)
        }
        return json.dumps(result)
    except Exception as e:
        return json.dumps({"status": "error", "message": str(e)})
```

### Step 3: Heterogeneous Orchestration Loop
We now combine both environments in a loop:
1. The **GPU (OpenRouter)** processes the user prompt, determines that a calculation tool is required, and emits a structured tool call.
2. The **CPU (Intel Xeon)** intercepts this call, executes the calculation locally with Pandas, and packages the result.
3. The **GPU (OpenRouter)** processes the tool output to construct the final text response.

```python
def run_heterogeneous_pipeline(user_prompt: str):
    # Define the tool signature for the LLM
    tools = [{
        "type": "function",
        "function": {
            "name": "cpu_calculate_financial_metrics",
            "description": "Calculates total revenue and average profit margin from a JSON array of financial periods.",
            "parameters": {
                "type": "object",
                "properties": {
                    "revenue_data_json": {
                        "type": "string",
                        "description": "JSON string array of objects. Example: '[{\"Revenue\": 100, \"Net_Profit\": 20}]'"
                    }
                },
                "required": ["revenue_data_json"]
            }
        }
    }]

    messages = [{"role": "user", "content": user_prompt}]

    print("--- PHASE 1: Prefill & Generation on Remote GPU (OpenRouter Nemotron) ---")
    response = client.chat.completions.create(
        model=MODEL_NAME,
        messages=messages,
        tools=tools,
        tool_choice="auto"
    )

    response_message = response.choices[0].message
    tool_calls = response_message.tool_calls

    if tool_calls:
        print("\n[GPU Output] Tool call detected! Delegating computation back to CPU...")
        messages.append(response_message)

        for tool_call in tool_calls:
            function_name = tool_call.function.name
            function_args = json.loads(tool_call.function.arguments)

            if function_name == "cpu_calculate_financial_metrics":
                print(f"\n--- PHASE 2: Local Execution on Intel Xeon CPU ---")
                # Execute locally on Colab Xeon CPU
                tool_output = cpu_calculate_financial_metrics(
                    revenue_data_json=function_args.get("revenue_data_json")
                )
                print(f"[CPU Output] Results computed: {tool_output}")

                messages.append({
                    "tool_call_id": tool_call.id,
                    "role": "tool",
                    "name": function_name,
                    "content": tool_output,
                })

        print("\n--- PHASE 3: Decoding Final Synthesis on Remote GPU ---")
        final_response = client.chat.completions.create(
            model=MODEL_NAME,
            messages=messages
        )
        print(f"\n[GPU Output] Final Response:\n{final_response.choices[0].message.content}")
    else:
        print(f"\n[GPU Output] Direct Response:\n{response_message.content}")
```

```python
import gradio as gr
import psutil
import platform
import time

def run_dashboard_pipeline(user_prompt: str):
    # Enforce Financial Analyst Agent Persona
    system_context = (
        "You are a specialized Heterogeneous Financial Analyst Agent. Your main job is to structure raw data "
        "into clean JSON objects and delegate analytical calculation requests to the local `cpu_calculate_financial_metrics` tool. "
        "Always prefer calling the available tool when financial datasets are present."
    )

    tools = [{
        "type": "function",
        "function": {
            "name": "cpu_calculate_financial_metrics",
            "description": "Calculates total revenue and average profit margin from a JSON array of financial periods.",
            "parameters": {
                "type": "object",
                "properties": {
                    "revenue_data_json": {
                        "type": "string",
                        "description": "JSON string array of objects. Example: '[{\"Revenue\": 100, \"Net_Profit\": 20}]'"
                    }
                },
                "required": ["revenue_data_json"]
            }
        }
    }]

    # Measure system metrics prior to run
    cpu_before = psutil.cpu_percent(interval=None)
    mem_info = psutil.virtual_memory()

    messages = [
        {"role": "system", "content": system_context},
        {"role": "user", "content": user_prompt}
    ]

    phase_logs = ""

    # --- Phase 1: GPU Prefill ---
    phase_logs += "[GPU] Dispatching Prompt to OpenRouter (Nemotron)...\n"
    try:
        response = client.chat.completions.create(
            model=MODEL_NAME,
            messages=messages,
            tools=tools,
            tool_choice="auto"
        )
    except Exception as e:
        return f"OpenRouter API Error: {str(e)}", "N/A", "N/A", "N/A"

    response_message = response.choices[0].message
    tool_calls = response_message.tool_calls

    # Grab token usage details if provided by OpenRouter
    usage = getattr(response, 'usage', None)
    prefill_tokens = usage.prompt_tokens if usage else "Unknown"
    decoding_tokens = usage.completion_tokens if usage else "Unknown"

    tool_results_log = "No local tool calls requested."

    if tool_calls:
        phase_logs += "[GPU] Tool Call emitted! Delegating processing to local CPU...\n"
        messages.append(response_message)

        for tool_call in tool_calls:
            if tool_call.function.name == "cpu_calculate_financial_metrics":
                # Measure active CPU spikes during the Pandas metric execution
                cpu_active = psutil.cpu_percent(interval=0.1)

                args = json.loads(tool_call.function.arguments)
                cpu_output = cpu_calculate_financial_metrics(args.get("revenue_data_json"))

                tool_results_log = f"Executed cpu_calculate_financial_metrics locally:\n{cpu_output}"
                phase_logs += f"[CPU] Pandas calculations processed successfully! (Xeon Load spiked to: {cpu_active}%)\n"

                messages.append({
                    "tool_call_id": tool_call.id,
                    "role": "tool",
                    "name": "cpu_calculate_financial_metrics",
                    "content": cpu_output,
                })

        # --- Phase 3: GPU final synthesis ---
        phase_logs += "[GPU] Synthesizing final analytical report...\n"
        final_response = client.chat.completions.create(
            model=MODEL_NAME,
            messages=messages
        )
        final_output = final_response.choices[0].message.content

        final_usage = getattr(final_response, 'usage', None)
        if final_usage:
            prefill_tokens = (prefill_tokens if isinstance(prefill_tokens, int) else 0) + final_usage.prompt_tokens
            decoding_tokens = (decoding_tokens if isinstance(decoding_tokens, int) else 0) + final_usage.completion_tokens
    else:
        final_output = response_message.content
        phase_logs += "[GPU] Completed Direct Synthesis.\n"

    # Post-execution metrics extraction
    cpu_after = psutil.cpu_percent(interval=None)
    cpu_cores = psutil.cpu_count(logical=True)

    system_specs = (
        f"Processor: {platform.processor()}\n"
        f"System: {platform.system()} {platform.release()}\n"
        f"Available Cores: {cpu_cores}\n"
        f"Memory Load: {mem_info.percent}% ({round(mem_info.used / (1024**3), 2)} GB used of {round(mem_info.total / (1024**3), 2)} GB)"
    )

    cpu_stats = f"Baseline: {cpu_before}% | Execution Spike: {cpu_after}%"
    token_stats = f"Prompt Prefill Tokens: {prefill_tokens}\nGeneration/Decoding Tokens: {decoding_tokens}"

    return final_output, tool_results_log, system_specs, cpu_stats, token_stats
```

```python
import gradio as gr
import matplotlib.pyplot as plt
import numpy as np
import time
import json
import psutil
import platform

def run_dashboard_pipeline_with_plot_stream(user_prompt: str):
    # Enforce Financial Analyst Agent Persona
    system_context = (
        "You are a specialized Heterogeneous Financial Analyst Agent. Your main job is to structure raw data "
        "into clean JSON objects and delegate analytical calculation requests to the local `cpu_calculate_financial_metrics` tool. "
        "Always prefer calling the available tool when financial datasets are present."
    )

    tools = [{
        "type": "function",
        "function": {
            "name": "cpu_calculate_financial_metrics",
            "description": "Calculates total revenue and average profit margin from a JSON array of financial periods.",
            "parameters": {
                "type": "object",
                "properties": {
                    "revenue_data_json": {
                        "type": "string",
                        "description": "JSON string array of objects. Example: '[{\"Revenue\": 100, \"Net_Profit\": 20}]'"
                    }
                },
                "required": ["revenue_data_json"]
            }
        }
    }]

    # Record CPU utilization timeline
    cpu_timeline = []
    time_steps = []

    # Baseline samples
    for i in range(5):
        cpu_timeline.append(psutil.cpu_percent(interval=0.05))
        time_steps.append(len(time_steps))

    messages = [
        {"role": "system", "content": system_context},
        {"role": "user", "content": user_prompt}
    ]

    # Yield initial state to make the UI immediately responsive
    yield "🔄 Contacting remote GPU for initial routing (Phase 1)...", "Evaluating...", "Gathering...", None, "Calculating..."

    try:
        response = client.chat.completions.create(
            model=MODEL_NAME,
            messages=messages,
            tools=tools,
            tool_choice="auto"
        )
    except Exception as e:
        yield f"OpenRouter API Error: {str(e)}", "N/A", "N/A", None, "N/A"
        return

    response_message = response.choices[0].message
    tool_calls = response_message.tool_calls

    usage = getattr(response, 'usage', None)
    prefill_tokens = usage.prompt_tokens if usage else "Unknown"
    decoding_tokens = usage.completion_tokens if usage else "Unknown"

    tool_results_log = "No local tool calls requested."

    if tool_calls:
        for tool_call in tool_calls:
            if tool_call.function.name == "cpu_calculate_financial_metrics":
                # Sample high-frequency CPU spikes during active local math processing
                for _ in range(8):
                    cpu_timeline.append(psutil.cpu_percent(interval=0.05))
                    time_steps.append(len(time_steps))

                args = json.loads(tool_call.function.arguments)
                cpu_output = cpu_calculate_financial_metrics(args.get("revenue_data_json"))
                tool_results_log = f"Executed cpu_calculate_financial_metrics locally:\n{cpu_output}"

                messages.append({
                    "tool_call_id": tool_call.id,
                    "role": "tool",
                    "name": "cpu_calculate_financial_metrics",
                    "content": cpu_output,
                })

        yield "🔄 Running local Xeon CPU tools (Phase 2)... Sending results back to GPU...", tool_results_log, "Gathering...", None, "Updating..."

        # --- Phase 3: GPU Streaming final synthesis ---
        try:
            stream_response = client.chat.completions.create(
                model=MODEL_NAME,
                messages=messages,
                stream=True
            )
        except Exception as e:
            yield f"OpenRouter API Error during streaming: {str(e)}", tool_results_log, "N/A", None, "N/A"
            return

        final_output = ""
        for chunk in stream_response:
            delta = chunk.choices[0].delta
            if hasattr(delta, 'content') and delta.content:
                final_output += delta.content

                # Re-generate current CPU plot frame
                fig, ax = plt.subplots(figsize=(6, 3))
                ax.plot(time_steps, cpu_timeline, color="#FF5733", linewidth=2.5, marker='o', label="CPU Load (%)")
                ax.fill_between(time_steps, cpu_timeline, color="#FF5733", alpha=0.15)
                ax.set_title("Local Intel Xeon CPU Utilization Profile", fontsize=10, fontweight='bold')
                ax.set_ylabel("CPU Usage (%)")
                ax.set_xlabel("Execution Pipeline Time Steps")
                ax.grid(True, linestyle="--", alpha=0.5)
                ax.set_ylim(0, max(max(cpu_timeline) + 15, 100))
                plt.tight_layout()

                mem_info = psutil.virtual_memory()
                system_specs = (
                    f"Processor: {platform.processor()}\n"
                    f"System: {platform.system()} {platform.release()}\n"
                    f"Available Cores: {psutil.cpu_count(logical=True)}\n"
                    f"Memory Load: {mem_info.percent}% ({round(mem_info.used / (1024**3), 2)} GB used of {round(mem_info.total / (1024**3), 2)} GB)"
                )
                token_stats = f"Prompt Prefill Tokens: {prefill_tokens}\nGeneration/Decoding Tokens: {decoding_tokens} (Streaming...)"

                yield final_output, tool_results_log, system_specs, fig, token_stats
                plt.close(fig)
    else:
        final_output = response_message.content

    # Post-execution idle samples
    for _ in range(5):
        cpu_timeline.append(psutil.cpu_percent(interval=0.05))
        time_steps.append(len(time_steps))

    # Generate final static plot
    fig, ax = plt.subplots(figsize=(6, 3))
    ax.plot(time_steps, cpu_timeline, color="#FF5733", linewidth=2.5, marker='o', label="CPU Load (%)")
    ax.fill_between(time_steps, cpu_timeline, color="#FF5733", alpha=0.15)
    ax.set_title("Local Intel Xeon CPU Utilization Spike Profile", fontsize=10, fontweight='bold')
    ax.set_ylabel("CPU Usage (%)")
    ax.set_xlabel("Execution Pipeline Time Steps")
    ax.grid(True, linestyle="--", alpha=0.5)
    ax.set_ylim(0, max(max(cpu_timeline) + 15, 100))
    plt.tight_layout()

    mem_info = psutil.virtual_memory()
    system_specs = (
        f"Processor: {platform.processor()}\n"
        f"System: {platform.system()} {platform.release()}\n"
        f"Available Cores: {psutil.cpu_count(logical=True)}\n"
        f"Memory Load: {mem_info.percent}% ({round(mem_info.used / (1024**3), 2)} GB used of {round(mem_info.total / (1024**3), 2)} GB)"
    )
    token_stats = f"Prompt Prefill Tokens: {prefill_tokens}\nGeneration/Decoding Tokens: {decoding_tokens} (Streaming Complete)"

    yield final_output, tool_results_log, system_specs, fig, token_stats
    plt.close(fig)

# Dynamic Web UI setup supporting streaming tokens
with gr.Blocks() as demo_visual_stream:
    gr.Markdown("# 🏦 Heterogeneous Financial Analyst Agent Dashboard (Streaming Mode)")
    gr.Markdown(
        "**Architecture Overview:** Heavy-lifting Reasoning & Decoding runs on the **Remote Cloud GPU (OpenRouter Nemotron)**, "
        "while Pandas calculations and data verification execute on the **Local Intel Xeon CPU** inside this Colab runtime."
    )

    with gr.Row():
        with gr.Column(scale=2):
            gr.Markdown("### 🛠️ Input Section")
            input_text = gr.Textbox(
                value=default_prompt,
                lines=10,
                label="Financial Dataset & Analyst Prompt",
                placeholder="Enter raw text containing financial quarterly records..."
            )
            submit_btn = gr.Button("🚀 Run Orchestration Loop", variant="primary")

        with gr.Column(scale=3):
            gr.Markdown("### 📈 Analysis Output")
            output_report = gr.Markdown(label="GPU Final Synthesized Report")
            output_tool = gr.Code(label="CPU Pandas Tool Result Payload", language="json")

    with gr.Row():
        with gr.Column():
            gr.Markdown("### 🖥️ Local Intel Xeon CPU Utilisation Profile")
            cpu_plot = gr.Plot(label="Active CPU Execution Spike Timeline")
            spec_box = gr.Textbox(label="Local System Specifications", lines=4, interactive=False)

        with gr.Column():
            gr.Markdown("### 📡 Remote GPU Token Utilisation")
            token_box = gr.Textbox(label="OpenRouter Token Tracking", lines=4, interactive=False)

    submit_btn.click(
        fn=run_dashboard_pipeline_with_plot_stream,
        inputs=[input_text],
        outputs=[output_report, output_tool, spec_box, cpu_plot, token_box]
    )

# Launch with share=True for clean visual hosting inside Colab iframe
demo_visual_stream.launch(share=True, debug=True)
```
